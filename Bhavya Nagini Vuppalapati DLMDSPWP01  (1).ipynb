# Step 2: Data Loader Implementation

```python
# Step 2: Data Loader Implementation
class DataLoader:
    def __init__(self, train_path, test_path, ideal_path):
        self.train_path = train_path
        self.test_path = test_path
        self.ideal_path = ideal_path

    def load_data(self):
        # Load training data
        train_data = pd.read_csv(self.train_path)
        print("Training data loaded successfully!")
        print("Training Data Head:")
        print(train_data.head())  # Display the first few rows of training data
        
        # Load test data
        test_data = pd.read_csv(self.test_path)
        print("\nTest data loaded successfully!")
        print("Test Data Head:")
        print(test_data.head())  # Display the first few rows of test data
        
        # Load ideal functions
        ideal_data = pd.read_csv(self.ideal_path)
        print("\nIdeal functions data loaded successfully!")
        print("Ideal Functions Data Head:")
        print(ideal_data.head())  # Display the first few rows of ideal functions data
        
        return train_data, test_data, ideal_data

# File paths for datasets
train_file = "train.csv"
test_file = "test.csv"
ideal_file = "ideal.csv"

# Instantiate DataLoader and load datasets
data_loader = DataLoader(train_file, test_file, ideal_file)
train_data, test_data, ideal_data = data_loader.load_data()
```

# Step 3: Database Manager Implementation

```python
from sqlalchemy.sql import text

class DatabaseManager:
    def __init__(self, db_name="assignment.db"):
        # Initialize the database connection
        self.engine = db.create_engine(f"sqlite:///{db_name}")
        self.connection = self.engine.connect()
        self.metadata = db.MetaData()
        print(f"Database '{db_name}' initialized successfully!")

    def create_table(self, table_name, dataframe):
        # Save a DataFrame to a database table
        dataframe.to_sql(table_name, self.engine, if_exists="replace", index=False)
        print(f"Table '{table_name}' created successfully!")

    def fetch_data(self, table_name):
        # Fetch data from a database table using SQLAlchemy text
        query = text(f"SELECT * FROM {table_name}")
        result = pd.read_sql(query, self.connection)
        print(f"Data fetched from table '{table_name}':")
        print(result.head())  # Display the first 5 rows
        return result

    def close_connection(self):
        # Close the database connection
        self.connection.close()
        print("Database connection closed.")

# Example usage
db_manager = DatabaseManager()

# Create tables for training, test, and ideal datasets
db_manager.create_table("train_data", train_data)
db_manager.create_table("test_data", test_data)
db_manager.create_table("ideal_functions", ideal_data)

# Fetch and preview data from the "train_data" table
fetched_data = db_manager.fetch_data("train_data")

```

```python
from sqlalchemy.sql import text

# Step 4: Querying the Database to Retrieve First Few Rows
def query_table_first_rows(db_manager, table_name, num_rows=5):
    """
    Query the database to retrieve the first few rows from the specified table.
    
    Parameters:
    - db_manager: Instance of DatabaseManager class
    - table_name: Name of the table to query
    - num_rows: Number of rows to fetch (default is 5)
    """
    # Convert the query into an SQLAlchemy text object
    query = text(f"SELECT * FROM {table_name} LIMIT {num_rows}")
    # Use db_manager.connection to execute the query
    data = pd.read_sql(query, db_manager.connection)
    print(f"First {num_rows} rows from table '{table_name}':")
    print(data)
    return data

# Example usage
# Query first 5 rows from each table
train_data_sample = query_table_first_rows(db_manager, "train_data")
test_data_sample = query_table_first_rows(db_manager, "test_data")
ideal_functions_sample = query_table_first_rows(db_manager, "ideal_functions")
```

# Step 5: Model Training - Least Squares Fit

```python
class LeastSquaresModel:
    def __init__(self, train_data, ideal_functions):
        """
        Initialize the Least Squares Model class.
        
        Parameters:
        - train_data: DataFrame containing training data with 'x' and 'y' columns.
        - ideal_functions: DataFrame containing ideal functions with 'x' and 'y' columns.
        """
        self.train_data = train_data
        self.ideal_functions = ideal_functions

    def calculate_least_squares(self):
        """
        Perform least squares fitting to find the best-fit ideal function for each training column.

        Returns:
        - Dictionary mapping each training column to its best-fit ideal function.
        """
        best_fits = {}
        for train_col in self.train_data.columns[1:]:  # Skip the 'x' column
            min_error = float('inf')
            best_function = None
            
            train_y = self.train_data[train_col]
            for ideal_col in self.ideal_functions.columns[1:]:  # Skip the 'x' column
                ideal_y = self.ideal_functions[ideal_col]
                
                # Calculate least squares error
                error = np.sum((train_y - ideal_y) ** 2)
                if error < min_error:
                    min_error = error
                    best_function = ideal_col
            
            # Store the best-fit function for the current training column
            best_fits[train_col] = best_function
        
        return best_fits

    def train(self):
        """
        Train the model by finding the best-fit ideal functions.

        Returns:
        - DataFrame containing the mapping of training columns to ideal functions.
        """
        best_fit_functions = self.calculate_least_squares()
        print("Best-fit ideal functions for each training column:")
        for train_col, ideal_col in best_fit_functions.items():
            print(f"{train_col} -> {ideal_col}")
        
        return best_fit_functions

# Example Usage
# Initialize the Least Squares Model with training and ideal function data
ls_model = LeastSquaresModel(train_data, ideal_data)

# Train the model to find the best-fit ideal functions
best_fit_functions = ls_model.train()
```

# Step 6: Test Data Mapping

```python
class TestDataMapper:
    def __init__(self, test_data, ideal_functions, best_fit_functions):
        """
        Initialize the TestDataMapper class.

        Parameters:
        - test_data: DataFrame containing the test dataset with 'x' and 'y' columns.
        - ideal_functions: DataFrame containing the ideal functions.
        - best_fit_functions: Dictionary mapping training columns to their best-fit ideal functions.
        """
        self.test_data = test_data
        self.ideal_functions = ideal_functions
        self.best_fit_functions = best_fit_functions

    def map_test_data(self, deviation_factor=np.sqrt(2)):
        """
        Map the test data to the best-fit ideal functions based on the deviation criteria.

        Parameters:
        - deviation_factor: Maximum allowed deviation factor (default is sqrt(2)).

        Returns:
        - DataFrame containing mapped test data with deviations and ideal functions.
        """
        mapped_results = []

        for _, test_row in self.test_data.iterrows():
            x_test, y_test = test_row["x"], test_row["y"]
            min_deviation = float("inf")
            best_function = None

            for ideal_col in self.best_fit_functions.values():
                ideal_y = self.ideal_functions.loc[self.ideal_functions["x"] == x_test, ideal_col].values[0]
                deviation = abs(y_test - ideal_y)

                # Check if the deviation is within the allowed range
                max_allowed_deviation = deviation_factor * max(
                    abs(self.ideal_functions[ideal_col] - self.test_data["y"].max())
                )
                if deviation < min_deviation and deviation <= max_allowed_deviation:
                    min_deviation = deviation
                    best_function = ideal_col

            # Store the mapping result
            mapped_results.append((x_test, y_test, best_function, min_deviation))

        # Create a DataFrame with the mapped results
        mapped_df = pd.DataFrame(
            mapped_results, columns=["x", "y", "ideal_function", "deviation"]
        )
        return mapped_df


# Example Usage
# Initialize the TestDataMapper with test data, ideal functions, and best-fit functions
mapper = TestDataMapper(test_data, ideal_data, best_fit_functions)

# Map the test data
mapped_test_data = mapper.map_test_data()

# Display the first few rows of the mapped data
print("Mapped Test Data:")
print(mapped_test_data.head())
```

# Test Data Mapped to Ideal Functions visualisation

```python
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import Legend
import pandas as pd
import numpy as np

# Enable Bokeh inline visualization
output_notebook()

def map_test_to_ideal(test_data, ideal_data, threshold=np.sqrt(2)):
    """
    Map each test data point to its corresponding ideal function based on minimum deviation.
    
    Parameters:
    - test_data: DataFrame containing test data with columns ['x', 'y'].
    - ideal_data: DataFrame containing ideal functions with columns ['x', 'f1', 'f2', ...].
    - threshold: Maximum allowable deviation for mapping (default is sqrt(2)).
    
    Returns:
    - test_data: Updated DataFrame with columns 'mapped_function' and 'deviation'.
    """
    # Initialize columns for mapped function and deviation
    test_data['mapped_function'] = None
    test_data['deviation'] = None

    for idx, row in test_data.iterrows():
        x = row['x']
        y_test = row['y']

        min_deviation = float('inf')
        best_function = None

        for ideal_function in ideal_data.columns[1:]:  # Skip the 'x' column
            # Get the ideal function value at the same x
            ideal_row = ideal_data.loc[ideal_data['x'] == x]
            if ideal_row.empty:
                continue  # Skip if x is not found in ideal_data
            y_ideal = ideal_row[ideal_function].values[0]
            deviation = abs(y_test - y_ideal)

            # Check if deviation is within threshold and update the best function
            if deviation < min_deviation and deviation <= threshold:
                min_deviation = deviation
                best_function = ideal_function

        # Update the mapped function and deviation for the test point
        test_data.at[idx, 'mapped_function'] = best_function
        test_data.at[idx, 'deviation'] = min_deviation

    return test_data

def visualize_test_data_mapping(test_data, ideal_data, ideal_function_column="mapped_function"):
    """
    Visualize test data points mapped to their corresponding ideal functions.
    
    Parameters:
    - test_data: DataFrame containing test data with columns ['x', 'y', ideal_function_column, 'deviation'].
    - ideal_data: DataFrame containing ideal functions with columns ['x', 'f1', 'f2', ...].
    - ideal_function_column: Name of the column in test_data that contains the mapped ideal function names.
    """
    # Filter out rows with None in the mapped_function column
    test_data_mapped = test_data[test_data[ideal_function_column].notnull()]

    # Create a Bokeh figure for visualization
    p = figure(title="Test Data Mapped to Ideal Functions",
               x_axis_label="x", y_axis_label="y", width=700, height=400)

    # Plot the ideal functions that test data points are mapped to
    for ideal_function in test_data_mapped[ideal_function_column].unique():
        if ideal_function not in ideal_data.columns:
            raise KeyError(f"Ideal function '{ideal_function}' not found in ideal_data.")
        p.line(ideal_data['x'], ideal_data[ideal_function],
               legend_label=f"Ideal {ideal_function}",
               line_width=2, line_dash="dashed", color="green")

    # Plot the test data points
    p.scatter(test_data_mapped['x'], test_data_mapped['y'], size=8, color="red", legend_label="Mapped Test Data")

    # Configure the legend
    p.legend.title = "Legend"
    p.legend.location = "top_left"
    p.legend.click_policy = "hide"

    # Show the plot
    show(p)

# Example Usage
# Load test_data and ideal_data
test_data = pd.read_csv("test.csv")
ideal_data = pd.read_csv("ideal.csv")

# Map test data to ideal functions
test_data = map_test_to_ideal(test_data, ideal_data)

# Visualize the mapping
visualize_test_data_mapping(test_data, ideal_data, ideal_function_column="mapped_function")
```

# Deviation Between Test Data and Ideal Functions

```python
from bokeh.plotting import figure, show, output_notebook
import pandas as pd
import numpy as np

# Enable Bokeh inline visualization
output_notebook()

def calculate_deviation(test_data, ideal_data):
    """
    Calculate the deviation between test data points and their mapped ideal functions.

    Parameters:
    - test_data: DataFrame containing test data with columns ['x', 'y'].
    - ideal_data: DataFrame containing ideal functions with columns ['x', 'f1', 'f2', ...].

    Returns:
    - test_data: Updated DataFrame with columns 'mapped_function' and 'deviation'.
    """
    # Initialize columns for mapped function and deviation
    test_data['mapped_function'] = None
    test_data['deviation'] = None

    for idx, row in test_data.iterrows():
        x = row['x']
        y_test = row['y']

        min_deviation = float('inf')  # Initialize with a high value
        best_function = None

        for ideal_function in ideal_data.columns[1:]:  # Skip the 'x' column
            # Find the corresponding ideal function value for the same 'x'
            ideal_row = ideal_data.loc[ideal_data['x'] == x]
            if ideal_row.empty:
                continue  # Skip if 'x' is not found in ideal_data
            y_ideal = ideal_row[ideal_function].values[0]
            deviation = abs(y_test - y_ideal)  # Calculate deviation

            # Check if this deviation is the smallest
            if deviation < min_deviation:
                min_deviation = deviation
                best_function = ideal_function

        # Update the test data with the best mapping and deviation
        test_data.at[idx, 'mapped_function'] = best_function
        test_data.at[idx, 'deviation'] = min_deviation

    return test_data

def visualize_deviation(test_data):
    """
    Visualize the deviation of test data points from their mapped ideal functions.

    Parameters:
    - test_data: DataFrame containing test data with columns ['x', 'y', 'mapped_function', 'deviation'].
    """
    # Filter test data with valid mapped functions
    test_data_mapped = test_data[test_data['mapped_function'].notnull()]

    # Create a Bokeh figure
    p = figure(title="Deviation of Test Data from Ideal Functions",
               x_axis_label="x", y_axis_label="Deviation",
               width=400, height=400)

    # Add a scatter plot for deviations
    p.scatter(test_data_mapped['x'], test_data_mapped['deviation'], size=8, color="red",
              legend_label="Deviation from Ideal Function")

    # Configure the legend
    p.legend.title = "Legend"
    p.legend.location = "top_left"

    # Show the plot
    show(p)

# Example Usage
# Load test_data and ideal_data
test_data = pd.read_csv("test.csv")
ideal_data = pd.read_csv("ideal.csv")

# Calculate deviations and map test data to ideal functions
test_data = calculate_deviation(test_data, ideal_data)

# Visualize the deviations
visualize_deviation(test_data)

# Display the updated test_data with 'mapped_function' and 'deviation'
print(test_data.head())
```
